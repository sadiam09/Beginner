```python
for column in range(1,6):
    print("o o o o o")
```

#### Solution

```python
n = 5
for i in range(n): # i is for each row 
    for j in range(n): # j is for each column
        print('o', end='  ')
    print()
```

#### Break

```python
for number in range(1, 6):  
    if number == 5:
        print("Stopping the loop!")
        break
    print(number)
    # Exits the loop when number is 5, stops the loop
```

#### Pass

```python
for number in range(1, 6):
    if number == 3:
        pass  # Placeholder; does nothing when number == 3
    else:
        print(number)
```

#### Continue

```python
for number in range(1, 6):
    if number == 3:
        continue  # goes to the next iteration
    print(number)
```

#### Matplot lib 

Not a built in library so needs to be downloaded in the terminal as `pip install matplot lib`
* Matplotlib is a comprehensive library for creating static, animated, and interactive visualizations in Python
* Seaborn is built on top of **Matplotlib**
* In fact, in the majority of the use cases in Data Science, we use a matplotlib submodule called **`pyplot`**, which is imported under the **alias `plt`**

#### Example

```python
import matplotlib.pyplot as plt  # plt is an abbreviation
import math

plt.style.use('_mpl-gallery') #setting the style of the graph

# create empty lists
x = []
y = []

# generate data using a for-loop
for i in range(0, 100):
    xi = i * 0.1        # x goes from 0 to 9.9
    yi = math.sin(xi)  # y = sin(x)

    x.append(xi)
    y.append(yi)
# append is how you add items to a list 

# plot
fig, ax = plt.subplots(figsize=(8, 4), dpi=120)

# ax.plot(x, y, 'o-', linewidth=2)
ax.scatter(x, y, marker='x')

ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_title("y = sin(x)")

plt.show()
# visualises on screen 
```
