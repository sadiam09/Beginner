#### Cluster prediction for an individual input

The new person (customer) is:
* Age: 30
* Annual income: 60k
* Spending score: 50

#### Create input for the new customer

```python
import pandas as pd
import joblib
```

```python
new_customer_df = pd.DataFrame([[30, 60, 50]], columns=['Age', 'Annual_Income', 'Spending_Score'])  # N.b.: _2D_array_
new_customer_df
```

#### Predict the cluster

```python
kmeans = joblib.load("../models/kmeans_v1.pkl")
kmeans
```

```python
cluster_label = kmeans.predict(new_customer_df)
print(f"The customer belongs to cluster: {cluster_label[0]}")
```

If you apply any preprocessing to your original dataset (like PCA, scaling ...), you should apply the same to the <b>new_customer_df</b> otherwise prediction will be wrong!!!!!

```python
distances = kmeans.transform(new_customer_df)
print("Distances to cluster centers:", distances)
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/sadiam09/m32895-tb2-2026/refs/heads/main/datasets/medical-charges.csv')
df.head() 
```

```python
df.describe()
```

```python
df.info()
```

```python
sns.pairplot(data=df[['age', 'bmi', 'charges']])
plt.show()
```

```python
X = df[['age', 'bmi']]
```

```python
from sklearn.cluster import KMeans
```

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)
```

```python
kmeans = KMeans(n_clusters=5, init='k-means++', max_iter=300, n_init=10, random_state=42)
y_kmeans = kmeans.fit_predict(X)
df['Cluster'] = y_kmeans
df.head()
```

```python
df["region"].unique()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X.iloc[:,0], X.iloc[:,1], c=y_kmeans, s=150, cmap='viridis')  # review slicing through iterables; 0, 1 stands for columns
centers = kmeans.cluster_centers_  # Retrieves coordinates of cluster centers
plt.scatter(centers[:,0], centers[:,1], c='red', s=200, alpha=.75, marker='X')
plt.xlabel("bmi")
plt.ylabel("age")
plt.title("Insurance")
plt.show()
```

```python
X = df[['age', 'bmi', 'charges']]
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', max_iter=300, n_init=10, random_state=42)
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of clusters")
plt.ylabel("WCSS")
plt.title("Cluster number optimisation by elbow")
plt.show()
```

```python
kmeans = KMeans(n_clusters=6, init='k-means++', max_iter=300, n_init=10, random_state=42)
y_kmeans = kmeans.fit_predict(X)

df['Cluster_Age_BMI'] = y_kmeans

df.head()
```
