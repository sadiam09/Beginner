### <u>Exercise 1:</u>

```python
""""
Prime number
"""

# Ask user to input number 
number = int(input("Enter a number: ")) # input() function is fed to the int() function

is_prime = True # assume the number is prime (boolean)

if number <= 1:
    is_prime = False # if the number is lower than or equal to 1 it is not prime 
else:
    for i in range(2, number): # returns all the numbers from 2 to the input
        if number % i == 0:  # gives remainder of the division and if its equal to zero then its not prime
            is_prime = False 
            break # stops loop from checking anymore numbers 
if is_prime:
    print(f"{number} is a prime number") # f formats variable into the string 
else:
    print(f"{number} is not a prime number")
```

### <u>Exercise 3:</u>

```python
"""
Pattern
"""

n = 5

# First triangle
for i in range(n, 0, -1):  # i is for each row
    for j in range(i):  # j is for each column
        print('*', end='  ')
    print()

# Second triangle
for i in range(2, n + 1):  # i is for each row
    for j in range(i):  # j is for each column
        print('*', end='  ')
    print()
```

### <u>Exercise 4:</u>

```python
"""
This function asks the user to enter a string
and prints the string in reverse (backwards)
using Python slicing.
"""

def reverse_string(): # defining a function
    word = input("Enter a string: ") # assigning the variable word to the string that the user inputs

    reversed_word = word [::-1] # using slicing to assign the variable reversed_word to reverse the string inputted

    print(f"Reversed string: {reversed_word}") # printing the reversed string ouput

reverse_string() # calling the function
```

### <u>Exercise 5:</u>

```python
"""
This function accepts any iterable
of numbers and returns the
minimum and maximum values.
"""

def get_min_max(numbers): # define function that accepts an iterable of numbers
    min_val = min(numbers) # find the smallest number in collection
    max_val = max(numbers) # find the largest number in collection
    return min_val, max_val # return both values 

min1,max1 = get_min_max([1, 12, 16, 32, 50]) # pass list as the argument
print(f"Min value is: {min1}")
print(f"Max value is: {max1}")

min2,max2 = get_min_max("15683") # pass string as the argument
print(f"Min value is: {min2}")
print(f"Max value is: {max2}")

min3,max3 = get_min_max((1, 17, 89, 2)) # pass tuple as the argument
print(f"Min value is: {min3}")
print(f"Max value is: {max3}")

min4,max4 = get_min_max({1, 3, 68, 7}) # pass set as the argument
print(f"Min value is: {min4}")
print(f"Max value is: {max4}")

min5,max5 = get_min_max(range(6)) # pass range as the argument
print(f"Min value is: {min5}")
print(f"Max value is: {max5}")
```

### <u>Exercise 6:</u>

```python
"""
Guess the Number Game
"""

import random as r # importing the random module as 'r'

secret_number = r.randint(1,10) # generating a random inentegr between 1 and 10

guess = 0 # variable for the user inputs

while guess != secret_number: # loop that continues to ask user input until the correct number is guessed
    guess = int(input("Guess the number (between 1 and 10): ")) # input() function is fed to the int() function
    
    if guess > secret_number: # checks if guess is higher than the secret number
        print("Too high! Try again.") # print hint
    
    elif guess < secret_number: # checks if guess is lower than the secret number
        print("Too low! Try again.") # print hint

    if guess == secret_number: # check if the guess matches the secret number
        break
    
print("Congratulations! You guessed the number.")
```

### <u>Exercise 7:</u>

```python
"""
Password Checker
""" 

password = "strawberry" # variable to assign the correct password

while True: # infinite loop that will keep asking for password
    user_input = input("Enter password: ")

    if user_input == password: # check if the entered password matches the correct one
        print("Access Granted!") # success message
        break # exits loop if the password is correct

    else: # if password is incorrect
        print("Incorrect password. Try again.") # prints error message and continues the loop
```
