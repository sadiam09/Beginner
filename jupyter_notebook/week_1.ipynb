#### Data types

```python
print(type("Hello, World!")) # string
print(type(42)) # integer
print(type(3.145)) # float which is a decimal number
print(type(1j)) # complex number
print(type(["egg", "bacon", "spam"])) # list
print(type(("egg", "bacon", "spam"))) # tuple, which is a unqiue list and cant have two of the same strings
print(type(range(6))) # range
print(type({"name" : "John", "age" : 80})) # dictionary
print(type({"egg", "bacon", "spam"})) # set 
print(type(True)) # booleon
print(isinstance(3.14, int)) # prints if the first parameter is an instance of the second parameter type
```

#### Comparison operators

```python
print('Hello, World!' == 'Hello, World!') # the double equal sign is a comparison operator for checking if two things are equal
print(2 != 2) # exclamation point with equal is asking if theyre not equal to each other
print([1,2]<[1,2,3]) # checking the total value of each list
print(float(2)>=int(2)) # both functions either turn into a float or integer and then checks if greater than or equal to 
print('a'<'A') #This is False as 'a' is Unicode 97 where 'A' is 65
```

#### Logical operators

```python
print(True and True) # booleon operations. and checks if both of them are true
print(True and False) # this will come up false because theyre not both true
print(True or False) # or menas if anything is true it will return true
print(not (4 < 5 and 4 < 10)) # not just make it the opposite of if its true or false
```
