### <u>Exercise 1:</u>

```python
""""
Prime number
"""

# Ask user to input number 
number = int(input("Enter a number: ")) # input() function is fed to the int() function

is_prime = True # assume the number is prime (boolean)

if number <= 1:
    is_prime = False # if the number is lower than or equal to 1 it is not prime 
else:
    for i in range(2, number): # returns all the numbers from 2 to the input
        if number % i == 0:  # gives remainder of the division and if its equal to zero then its not prime
            is_prime = False 
            break # stops loop from checking anymore numbers 
if is_prime:
    print(f"{number} is a prime number") # f formats variable into the string 
else:
    print(f"{number} is not a prime number")
```

```python
"""
Pattern
"""

n = 5

# First triangle
for i in range(n, 0, -1):  # i is for each row
    for j in range(i):  # j is for each column
        print('*', end='  ')
    print()

# Second triangle
for i in range(2, n + 1):  # i is for each row
    for j in range(i):  # j is for each column
        print('*', end='  ')
    print()
```

```python
"""
This function asks the user to enter a string
and prints the string in reverse
using Python slicing.
"""

def reverse_string(): # defining a function
    word = input("Enter a string: ") # assigning the variable word to the string that the user inputs

    reversed_word = word [::-1] # using slicing to assign the variable reversed_word to reverse the string inputted

    print(f"Reversed string: {reversed_word}") # printing the reversed string ouput

reverse_string() # calling the function
```

### <u>Exercise 5:</u>

```python
"""
This function accepts any iterable
of numbers and returns the
minimum and maximum values.
"""

def get_min_max(numbers): # define function that accepts an iterable of numbers
    min_val = min(numbers) # find the smallest number in collection
    max_val = max(numbers) # find the largest number in collection
    return min_val, max_val # return both values 

min1,max1 = get_min_max([1, 12, 16, 32, 50]) # pass list as the argument
print(f"Min value is: {min1}")
print(f"Max value is: {max1}")

min2,max2 = get_min_max("15683") # pass string as the argument
print(f"Min value is: {min2}")
print(f"Max value is: {max2}")

min3,max3 = get_min_max((1, 17, 89, 2)) # pass tuple as the argument
print(f"Min value is: {min3}")
print(f"Max value is: {max3}")

min4,max4 = get_min_max({1, 3, 68, 7}) # pass set as the argument
print(f"Min value is: {min4}")
print(f"Max value is: {max4}")

min5,max5 = get_min_max(range(6)) # pass range as the argument
print(f"Min value is: {min5}")
print(f"Max value is: {max5}")
```

### <u>Exercise 6:</u>

```python
"""
Guess the Number Game
"""

import random as r # importing the random module as 'r'

secret_number = r.randint(1,10) # generating a random inentegr between 1 and 10

while True: # infinite loop that continues to ask user input until the correct number is guessed
    guess = int(input("Guess the number (between 1 and 10): ")) # input() function is fed to the int() function
    
    if guess > secret_number: # checks if guess is higher than the secret number
        print("Too high! Try again.") # print hint
    
    elif guess < secret_number: # checks if guess is lower than the secret number
        print("Too low! Try again.") # print hint

    if guess == secret_number: # check if the guess matches the secret number
        break
    
print("Congratulations! You guessed the number.")
```

### <u>Exercise 8:</u>

```python
"""
This function calculates the value of the series:

x*5^2/(1+2!) + x*25^2/(2+3!) ... N
"""

import math as m # importing math module as m

def calculate(x, N): # defining a function that calculates the equation 
    base_number = 5 # starting base number used in the series 
    total = 0 # variable to store the sum

    for i in range(N): # loop through N term of the series (defined by user input)
        next_base_number = base_number ** 2 #square base term for use in this term and the next term
        numerator = x * next_base_number # calculating numerator
        denominator = (i + 1) + m.factorial(i + 2) # calculate denominator using factorial (i starts at zero so need to add 1)
        result = numerator/denominator # calculate value of current term
        total += result # add term to total
        base_number = next_base_number # set base number for next term

    return total # returns final sum


x = int(input("Enter a value for x:")) # assigning the variable word to the string that the user inputs
N = int(input("Enter a value for N:"))

result = calculate(x, N) # calling the function with the inputs
print(f"Result: {result}") # print the result
```

### <u>Exercise 10:</u>

```python
"""
temperature data 
into numpy array
"""

import numpy as np  # import numpy library

temperature_data = [22,24,19,21,23,25,26,27,24,23,20,22,21,19,18,22,23,24,26,28,30,27,25,26,24,22,21,20,19,23] # temperature values in celsius
temperature_c = np.array(temperature_data) # convert list to numpy array

minimum = temperature_c.min() # find minimum temp
maximum = temperature_c.max() # find maximum temp
average = temperature_c.mean() # find average

print(f"Min temp: {minimum}") # print outputs
print(f"Max temp: {maximum}")
print(f"Average temp: {average}")
```

```python
# F = C * 9/5 + 32 (convert celcius to farenheight equation)
# using the https://numpy.org/doc/stable/user/quickstart.html to see how to multiply every element in an array
temperature_f = temperature_c * 9/5 + 32

print(f"Farenheight: {temperature_f}")
```

```python
temperature_above_25C = temperature_c[temperature_c > 25] # select temperatures greater than 25
temperature_below_20C = temperature_c[temperature_c < 20] # select temperatures less than 20

below_20C_count = temperature_below_20C.size # count number of temperatures in list below 20

print(f"Above 25C: {temperature_above_25C}") # display outputs
print(f"Below 20C count: {below_20C_count}")
```

```python
temp_c_sorted = np.sort(temperature_c) # sort temperatures in ascending order

print(f"Sorted celcius: {temp_c_sorted}") # display outputs
```

```python
reshape = temperature_c.reshape(5, 6) # reshape temp array into a 5x6 matrix

print(f"Reshaped: {reshape}")
```

### <u>Exercise 15:</u>

```python
"""
Titanic survival prediction
using a decision tree classifier
"""

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier      # decision tree model
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.tree import plot_tree                   # visualise decision tree
import pickle # save and load trained model
import matplotlib.pyplot as plt
import numpy as np
import itertools
from sklearn.preprocessing import OneHotEncoder      # convert categorical variables into a numerical format

def plot_confusion_matrix(cm, classes, # plots confusion matrix
                          normalize=False,
                          title='Confusion matrix',
                          cmap=plt.cm.Blues):
    """
    cm - confusion matrix
    classes - class list
    normalize - normalize to 1 if True
    title - plot title
    cmap - color map
    """

    if normalize:    # normalise values if required
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        print("Normalized confusion matrix")
    else:
        print('Confusion matrix, without normalization')

    plt.imshow(cm, interpolation='nearest', cmap=cmap) # plot matrix as image
    plt.title(title)
    plt.colorbar()

    tick_marks = np.arange(len(classes)) # label axis
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)

    fmt = '.2f' if normalize else 'd' # display values inside confusion matrix cells
    thresh = cm.max() / 2.

    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], fmt),
                 horizontalalignment="center",
                 color="white" if cm[i, j] > thresh else "black")

    plt.tight_layout()
    plt.ylabel('True label')
    plt.xlabel('Predicted label')

def experiment(max_depth, min_samples_split):

    """
    Builds and trains Decision Tree model
    """

    model = DecisionTreeClassifier(max_depth=max_depth, min_samples_split=min_samples_split, random_state=42)  # create decision tree model with chosen parameters
    model.fit(train.drop('Survived', axis=1), train['Survived']) #train model using training data

    preds = model.predict(test.drop('Survived', axis=1)) # prerdict survival on test data
    acc = accuracy_score(test['Survived'], preds) # Calculate accuracy metrics
    cm = confusion_matrix(test['Survived'], preds)

    print("accuracy", acc)

    plot_confusion_matrix(cm, classes=['Not Survived', 'Survived']) # Plot confusion matrix

    report = classification_report(test['Survived'], preds, target_names=['Not Survived', 'Survived']) # display classification report
    print(report)

    with open('model_dt.pkl', 'wb') as f:
        pickle.dump(model, f) # Save trained model in pickle format#############

df = pd.read_csv('https://raw.githubusercontent.com/Koldim2001/test_api/refs/heads/main/titanic.csv' ) # load titanic dataset
df = df[['Survived', 'Pclass', 'Age', 'Fare', 'SibSp', 'Parch', 'Sex', 'Embarked']] # select relevant features for the model
df = df.dropna() # remove rows with missing values

# https://www.datacamp.com/tutorial/one-hot-encoding-python-tutorial
enc = OneHotEncoder(sparse_output=False, handle_unknown='ignore') # Sex and Embarked are categorical variables so they must be converted to numbers

enc.fit(df[['Sex', 'Embarked']]) # fit encoder to dataset

one_hot = enc.transform(df[['Sex', 'Embarked']]) # transform categorical to encoded numerical columns

encoded_cols = enc.get_feature_names_out(['Sex', 'Embarked']) # get names of encoded columns

one_hot_df = pd.DataFrame(one_hot, columns=encoded_cols, index=df.index) # convert encoded array to dataframe

df = pd.concat([df, one_hot_df], axis=1) # combine encoded columns wiht original data

df = df.drop(columns=['Sex', 'Embarked']) # remove original text columns after encoding

train, test = train_test_split(df, test_size=0.2, random_state=42) # split data into training and testing

max_depth = 5
min_samples_split = 150 # Define model hyperparameters

experiment(max_depth, min_samples_split) # train and evluate model

with open('model_dt.pkl', 'rb') as f:
    model = pickle.load(f) # load saved model

person = pd.DataFrame({    # Predict outcome of Titanic trip for a person
    'Pclass': [3],
    'Age': [55],
    'Fare': [7.25],
    'SibSp': [0],
    'Parch': [0],
    'Sex_female': [0.0],
    'Sex_male': [1.0],
    'Embarked_C': [0.0],
    'Embarked_Q': [0.0],
    'Embarked_S': [1.0],
})

prediction = model.predict(person)
print(f"The model predicts {prediction}")

if prediction[0] == 1:
    print ("This person is most likely a survivor.")
else:
    print("This person is most likely deceased.") # interpret prediction

plt.figure(figsize=(80,20))
plot_tree(model, feature_names=df.drop(columns='Survived').columns, max_depth=5, filled=True) # visualise trained decision tree

importances = pd.Series(model.feature_importances_, index=df.drop(columns='Survived').columns)
importances = importances.sort_values(ascending=False) # sort features by importance

plt.figure(figsize=(10, 6)) # plot feature important bar chart
importances.plot(kind='bar')
plt.title('Feature Importance')
plt.ylabel('Importance')
plt.tight_layout()
plt.show()
plt.close()
```
