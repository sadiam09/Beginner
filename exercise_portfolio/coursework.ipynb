```python
""""
Prime number
"""

# Ask user to input number 
number = int(input("Enter a number: ")) # input() function is fed to the int() function

is_prime = True # assume the number is prime (boolean)

if number <= 1:
    is_prime = False # if the number is lower than or equal to 1 it is not prime 
else:
    for i in range(2, number): # returns all the numbers from 2 to the input
        if number % i == 0:  # gives remainder of the division and if its equal to zero then its not prime
            is_prime = False 
            break # stops loop from checking anymore numbers 
if is_prime:
    print(f"{number} is a prime number") # f formats variable into the string 
else:
    print(f"{number} is not a prime number")
```

### <u>Exercise 5:</u>

```python
"""
This function accepts any iterable
of numbers and returns the
minimum and maximum values.
"""

def get_min_max(numbers): # define function that accepts an iterable of numbers
    min_val = min(numbers) # find the smallest number in collection
    max_val = max(numbers) # find the largest number in collection
    return min_val, max_val # return both values 

min1,max1 = get_min_max([1, 12, 16, 32, 50]) # pass list as the argument
print(f"Min value is: {min1}")
print(f"Max value is: {max1}")

min2,max2 = get_min_max("15683") # pass string as the argument
print(f"Min value is: {min2}")
print(f"Max value is: {max2}")

min3,max3 = get_min_max((1, 17, 89, 2)) # pass tuple as the argument
print(f"Min value is: {min3}")
print(f"Max value is: {max3}")

min4,max4 = get_min_max({1, 3, 68, 7}) # pass set as the argument
print(f"Min value is: {min4}")
print(f"Max value is: {max4}")

min5,max5 = get_min_max(range(6)) # pass range as the argument
print(f"Min value is: {min5}")
print(f"Max value is: {max5}")
```

### <u>Exercise 6:</u>

```python
"""
Guess the Number Game
"""

import random as r # importing the random module as 'r'

secret_number = r.randint(1,10) # generating a random inentegr between 1 and 10

while True: # infinite loop that continues to ask user input until the correct number is guessed
    guess = int(input("Guess the number (between 1 and 10): ")) # input() function is fed to the int() function
    
    if guess > secret_number: # checks if guess is higher than the secret number
        print("Too high! Try again.") # print hint
    
    elif guess < secret_number: # checks if guess is lower than the secret number
        print("Too low! Try again.") # print hint

    if guess == secret_number: # check if the guess matches the secret number
        break
    
print("Congratulations! You guessed the number.")
```

### <u>Exercise 8:</u>

```python
"""
This function calculates the value of the series:

x*5^2/(1+2!) + x*25^2/(2+3!) ... N
"""

import math as m # importing math module as m

def calculate(x, N): # defining a function that calculates the equation 
    base_number = 5 # starting base number used in the series 
    total = 0 # variable to store the sum

    for i in range(N): # loop through N term of the series (defined by user input)
        next_base_number = base_number ** 2 #square base term for use in this term and the next term
        numerator = x * next_base_number # calculating numerator
        denominator = (i + 1) + m.factorial(i + 2) # calculate denominator using factorial (i starts at zero so need to add 1)
        result = numerator/denominator # calculate value of current term
        total += result # add term to total
        base_number = next_base_number # set base number for next term

    return total # returns final sum


x = int(input("Enter a value for x:")) # assigning the variable word to the string that the user inputs
N = int(input("Enter a value for N:"))

result = calculate(x, N) # calling the function with the inputs
print(f"Result: {result}") # print the result
```

### <u>Exercise 15:</u>

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.tree import plot_tree, export_text
import pickle
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import itertools
from sklearn.preprocessing import OneHotEncoder

def plot_confusion_matrix(cm, classes,
                          normalize=False,
                          title='Confusion matrix',
                          cmap=plt.cm.Blues):
    """
    Plots confusion matrix
    cm - confusion matrix
    classes - class list
    normalize - normalize to 1 if True
    title - plot title
    cmap - color map
    """

    if normalize:
        cm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
        print("Normalized confusion matrix")
    else:
        print('Confusion matrix, without normalization')

    plt.imshow(cm, interpolation='nearest', cmap=cmap)
    plt.title(title)
    plt.colorbar()
    tick_marks = np.arange(len(classes))
    plt.xticks(tick_marks, classes, rotation=45)
    plt.yticks(tick_marks, classes)

    fmt = '.2f' if normalize else 'd'
    thresh = cm.max() / 2.
    for i, j in itertools.product(range(cm.shape[0]), range(cm.shape[1])):
        plt.text(j, i, format(cm[i, j], fmt),
                 horizontalalignment="center",
                 color="white" if cm[i, j] > thresh else "black")

    plt.tight_layout()
    plt.ylabel('True label')
    plt.xlabel('Predicted label')

def experiment(max_depth, min_samples_split):
    """
    Builds and trains Decision Tree model
    """
    # Build and train Decision Tree model
    model = DecisionTreeClassifier(max_depth=max_depth, min_samples_split=min_samples_split, random_state=42)
    model.fit(train.drop('Survived', axis=1), train['Survived'])

    # Calculate accuracy metrics
    preds = model.predict(test.drop('Survived', axis=1))
    acc = accuracy_score(test['Survived'], preds)
    cm = confusion_matrix(test['Survived'], preds)

    print("accuracy", acc)

    # Plot confusion matrix
    plot_confusion_matrix(cm, classes=['Not Survived', 'Survived'])

    # Classification report
    report = classification_report(test['Survived'], preds, target_names=['Not Survived', 'Survived'])
    print(report)

    # Save model in pickle format
    with open('../models/model_dt.pkl', 'wb') as f:
        pickle.dump(model, f)

df = pd.read_csv('https://raw.githubusercontent.com/Koldim2001/test_api/refs/heads/main/titanic.csv' )
df = df[['Survived', 'Pclass', 'Age', 'Fare', 'SibSp', 'Parch', 'Sex', 'Embarked']]
df = df.dropna()

# https://www.datacamp.com/tutorial/one-hot-encoding-python-tutorial
enc = OneHotEncoder(sparse_output=False, handle_unknown='ignore')

enc.fit(df[['Sex', 'Embarked']])

one_hot = enc.transform(df[['Sex', 'Embarked']])

encoded_cols = enc.get_feature_names_out(['Sex', 'Embarked'])

one_hot_df = pd.DataFrame(one_hot, columns=encoded_cols, index=df.index)

df = pd.concat([df, one_hot_df], axis=1)

# remove original text columns after encoding
df = df.drop(columns=['Sex', 'Embarked'])

train, test = train_test_split(df, test_size=0.2)

# Define model hyperparameters
max_depth = 5
min_samples_split = 150

experiment(max_depth, min_samples_split)

with open('../models/model_dt.pkl', 'rb') as f:
    model = pickle.load(f)

# Predict outcome of Titanic trip for a person
person = pd.DataFrame({
    'Pclass': [3],
    'Age': [55],
    'Fare': [7.25],
    'SibSp': [0],
    'Parch': [0],
    'Sex_female': [0.0],
    'Sex_male': [1.0],
    'Embarked_C': [0.0],
    'Embarked_Q': [0.0],
    'Embarked_S': [1.0],
})

prediction = model.predict(person)
print(f"The model predicts {prediction}")

if prediction == [1]:
    print ("This person is, the most likely, is a survivor.")
else:
    print("This person, the most likely, perished.")

plt.figure(figsize=(80,20))
plot_tree(model, feature_names=df.drop(columns='Survived').columns, max_depth=5, filled=True)

importances = pd.Series(model.feature_importances_, index=df.drop(columns='Survived').columns)
importances = importances.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
importances.plot(kind='bar')
plt.title('Feature Importance')
plt.ylabel('Importance')
plt.tight_layout()
plt.show()
plt.close()
```
